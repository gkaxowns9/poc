- 퍼셉트론 : 다수의 신호를 입력으로 받아 하나의 신호를 출력하는 알고리즘 (0 또는 1을 신호로 가짐)

![image.png](./perceptron.png)


- 위 그림의 원 : 노드, 뉴런
- w : 가중치

- 뉴런의 신호에 가중치를 곱한 값을 전부 더한 값이 임계값 $\theta$ 를 넘어가면 1을 출력하며, 활성화 되었다고 한다

$$
y = \begin{cases} 0 & (w_1x_1 + w_2x_2 \le \theta) \\ 1 & (w_1x_1 + w_2x_2 > \theta) \end{cases}
$$

```python
import numpy as np

def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7

    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(AND(1, 0))
print(AND(0, 1))
print(AND(0, 0))
print(AND(1, 1))
```

```python
import numpy as np

def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2

    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(OR(1, 0))
print(OR(0, 1))
print(OR(0, 0))
print(OR(1, 1))
```

```python
import numpy as np

def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7

    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(NAND(1, 0))
print(NAND(0, 1))
print(NAND(0, 0))
print(NAND(1, 1))
```

- 선 하나로 값을 구분할 수 있는 OR, AND, NAND와는 달리, XOR의 경우는 불가능

- 다층 퍼셉트론 : 여러 퍼셉트론 층을 쌓아 

![image](./layer-xor.png)

```python
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    return AND(s1, s2)
    
print(XOR(1, 0))
print(XOR(0, 1))
print(XOR(0, 0))
print(XOR(1, 1))
```
