- 직접 가중치와 bias를 정하던 퍼셉트론과는 달리, 신경망은 데이터로부터 자동으로 가중치 매개변수의 적절한 값을 찾을 수 있음

### 활성화 함수
- 입력 신호의 총합을 출력 신호로 변환 (활성화를 일으키는 지 정하는 역할)


```python
import numpy as np
import matplotlib.pylab as plt

x = np.arange(-5.0, 5.0, 0.1) # -5.0부터 5.0까지 0.1씩
plt.ylim(-0.1, 1.1)
```

### Step Function

$$
h(x) = \begin{cases} 0 & (x \le 0) \\ 1 & (x > 0) \end{cases}
$$

```python
def step(x):
    y = x > 0
    return y.astype(np.int16)

y = step(x)
plt.plot(x, y)
plt.show()
```

### Sigmoid Function

$$
h(x) = \frac{1}{1 + \exp(-x)}
$$

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

y = sigmoid(x)
plt.plot(x, y)
plt.show()
```

- 신경망의 활성화 함수인 sigmoid와 퍼셉트론에서 관찰했던 step function의 큰 차이는 **매끄러움**
- 0 또는 1을 보여주는 퍼셉트론과 달리 꽤 연속적인 실수를 보여줌

- 미분 가능성과의 연관으로 보임

### ReLU Function (Rectified Linear Unit)
- 정류된 선형 유닛 (0 미만의 값을 없애서 이런 이름인 것 같음)

$$
h(x) = \begin{cases} x & (x > 0) \\ 0 & (x \leq 0) \end{cases}
$$

```python
def relu(x):
    return np.maximum(0, x) # 두 매개변수 중 더 큰 값을 반환

y = relu(x)
plt.plot(x, y)
plt.show()
```
